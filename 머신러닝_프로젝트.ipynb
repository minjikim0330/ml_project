```python
import random

def recommend_clothes(temp):
    if temp <= 4:
        tops = ["패딩", "두꺼운 코트", "니트", "기모 옷"]
        bottoms = ["두꺼운 바지"]

    elif 5 <= temp <= 8:
        tops = ["코트", "니트", "히트텍"]
        bottoms = ["청바지", "레깅스"]

    elif 9 <= temp <= 11:
        tops = ["트랜치코트", "야상", "니트"]
        bottoms = ["스타킹"]

    elif 12 <= temp <= 16:
        tops = ["자켓", "야상", "맨투맨", "니트"]
        bottoms = ["스타킹"]

    elif 17 <= temp <= 19:
        tops = ["얇은 니트", "맨투맨", "얇은 재킷", "가디건"]
        bottoms = ["청바지"]

    elif 20 <= temp <= 22:
        tops = ["얇은 가디건", "긴팔티", "얇은 셔츠"]
        bottoms = ["면바지", "청바지"]

    elif 23 <= temp <= 27:
        tops = ["반팔", "얇은 셔츠"]
        bottoms = ["반바지", "면바지"]

    else:
        tops = ["민소매", "반팔"]
        bottoms = ["반바지", "치마"]

    top = random.choice(tops)
    bottom = random.choice(bottoms)

    return top, bottom
```

```python
temp = int(input("현재 기온을 입력하세요: "))
top, bottom = recommend_clothes(temp)

print("👕 상의 추천:", top)
print("👖 하의 추천:", bottom)
```
